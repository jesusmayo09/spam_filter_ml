```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, confusion_matrix,classification_report
```

```python
# 1. Crear el dataset de ejemplo (correos)

datos = {
    'texto' : [
        'Gana dinero rapido trabajando desde casa',
        'Hola Juan, te adjunto el reporte de ventas',
        'Felicidades, ganaste un iPhone 14. Haz cliic aqui',
        'Reunion de equipo manana a las 10 am',
        'Oferta exclusiva, 50% de descuento solo por hoy',
        'Vamos a almorzar mas tarde?',
        'Aumenta tus ingresos sin esfuerzo invirtiendo ya',
        'Por favor revisa el documento para la presentacion',
    ],
    'etiqueta' : ['spam', 'ham', 'spam', 'ham', 'spam', 'ham', 'spam', 'ham']
}

# Convertir a DataFrame
df = pd.DataFrame(datos)
print("Vista de los datos de entrenamiento:")
print(df)
```

```python
# 2. Inicializar el vectorizador

vectorizer = CountVectorizer(stop_words=['de', 'la', 'el', 'a', 'para'])

# Convertir los textos a una matrix de conteo de palabras
X = vectorizer.fit_transform(df['texto'])
y = df['etiqueta']

print(f"\nSe encontraron {len(vectorizer.get_feature_names_out())} palarbas unicas.")
print("Vocabulario:", vectorizer.get_feature_names_out())

```

```python
# 3. Dividir datos (70% entrenamiento, 30% validacion)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 4. Inicializar y entrenar el modelo
modelo = MultinomialNB()
modelo.fit(X_train, y_train)

y_pred = modelo.predict_proba(X_test)
```

```python
# 6. Correos nuevos introducidos por el usuario

correos_nuevos = [
    "Urgente, haz clic aqui para reclamar tu premio exclusivo",
    "Hola profe, le envio la terea de inteligencia artificial",
    "Gana dinero rapido con este descuento",
    "Urgente y exclusivo, el costo de la conferencia baja un 0%",
    "Hola profe, le envio la terea de como funciona la oferta exclusiva",
    "Reunión sobre cómo ganar dinero"
]

# IMPORTANTE: Usar transform() y NO fit_transform() para datos nuevos
correos_vectorizados = vectorizer.transform(correos_nuevos)

# Predecir
predicciones = modelo.predict_proba(correos_vectorizados)

print("\n--- RESULTADOS DEL FILTRO ANTI-SPAM ---")
for correo,  prediccion in zip (correos_nuevos, predicciones):
  # Asumiendo que modelo.classes_ es ['ham', 'spam'], entonces prediccion[0] es prob_ham y prediccion[1] es prob_spam
  ham_prob = prediccion[0]
  spam_prob = prediccion[1]
  print(f"[HAM: {ham_prob:.2f}, SPAM: {spam_prob:.2f}] => {correo}")
```
